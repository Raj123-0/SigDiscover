# 04. COSMIC Assignment

This notebook demonstrates how to assign extracted signatures to COSMIC.

```python
from sigdiscover.assignment.similarity import assign_to_cosmic
import numpy as np

S_disc = np.random.dirichlet(np.ones(96), size=2)
S_cosmic = np.random.dirichlet(np.ones(96), size=10)
names = [f"SBS{i+1}" for i in range(10)]

assignments = assign_to_cosmic(S_disc, S_cosmic, names)
print(assignments)
```
