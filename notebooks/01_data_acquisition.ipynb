# 01. Data Acquisition

This notebook demonstrates how to download and load data for SigDiscover.

```python
from sigdiscover.data.download import download_cosmic_signatures
from sigdiscover.data.loaders import load_cosmic_signatures

# Note: downloading may take a few moments
download_cosmic_signatures(version="3.4", output_dir="../data/cosmic")
signatures, names = load_cosmic_signatures("SBS96", data_dir="../data/cosmic")
print(f"Loaded {len(names)} COSMIC signatures")
```
