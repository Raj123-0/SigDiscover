# 03. Signature Extraction

This notebook demonstrates how to extract signatures using NMF.

```python
from sigdiscover.utils.io import load_matrix
from sigdiscover.extraction.nmf import nmf_mutational_signatures
from sigdiscover.extraction.rank_selection import select_optimal_rank

M_df = load_matrix("../tests/test_data/mini_matrix.tsv")
M = M_df.values

res = select_optimal_rank(M, min_k=1, max_k=3, n_replicates=5)
print("Optimal K:", res['optimal_k'])

S, A, error = nmf_mutational_signatures(M, n_signatures=res['optimal_k'])
print("Signatures shape:", S.shape)
```
